```python
def least_squares_fit(x, y, order):
    """Solve normal equations: c = (V^T V)^{-1} V^T y."""
    V = build_vandermonde(x, order)
    VtV = V.T @ V
    Vty = V.T @ y
    coeffs = np.linalg.solve(VtV, Vty)  # more stable than explicit inverse
    return coeffs     

def build_vandermonde(x, order):
    """Build Vandermonde matrix: columns are x^0, x^1, ..., x^order."""
    return np.column_stack([x**i for i in range(order + 1)])

def eval_poly(x, coeffs):
    """Evaluate polynomial with ascending-order coefficients."""
    return sum(c * x**i for i, c in enumerate(coeffs))

def mle_analytic(x, y, order):
    V = build_vandermonde(x, order)
    VtV = V.T @ V
    Vty = V.T @ y
    coeffs = np.linalg.solve(VtV, Vty)

    residuals = y - V @ coeffs
    n = len(y)
    sigma2_mle = np.sum(residuals**2) / n   # MLE estimate of sigma² (note: biased)

    p = order + 1
    sigma2_unbiased = np.sum((y - V @ coeffs)**2) / ((len(y) - p) + 1e-9)
    return V, coeffs, sigma2_unbiased

def bayesian_approach(x, y, order, alpha, sigma):
    V = build_vandermonde(x, order)
    lambda_reg = sigma**2/alpha**2
    S_inv = (1/sigma**2) * (V.T @ V) + (1/alpha**2) * np.eye(order + 1)
    S = np.linalg.inv(S_inv)
    # print("Posterior covariance:")
    # print(S)
    m = (1/sigma**2) * S @ V.T @ y
    # print("Mean:")
    # print(m)
    A_map  = V.T @ V  +  lambda_reg * np.eye(order + 1)
    c_map  = np.linalg.solve(A_map, V.T @ y)     # shape: (P,)
    return m, S, V, c_map

def compute_posterior(x, y, order, alpha, sigma):
    V = build_vandermonde(x, order)
    S_inv = (1/sigma**2) * (V.T @ V) + (1/alpha**2) * np.eye(order + 1)
    S = np.linalg.inv(S_inv)
    # print("Posterior covariance:")
    # print(S)
    m = (1/sigma**2) * S @ V.T @ y
    # print("Mean:")
    # print(m)

    return m, S, V

def log_marginal_likelihood(N, x, y, order, alpha, sigma):
    N = len(y)
    p = order + 1
    m, S, V = compute_posterior(x, y, order, alpha, sigma)

    C = sigma**2 * np.eye(N)  +  alpha**2 * (V @ V.T)
    sign, log_det_C = np.linalg.slogdet(C)
    C_inv = np.linalg.inv(C)
    quadratic = y @ C_inv @ y
    log_ml = -0.5 * N * np.log(2 * np.pi) \
             -0.5 * log_det_C \
             -0.5 * quadratic

    return log_ml, m, S, V
```

#### Question 1
Least squares approach to curve fitting -> determine the optimal weight vector

For a polynomial of order n, we want to find coefficients $\mathbf{c} = [c_0, c_1, ..., c_n]$ such that: 

$y\approx c_0 + c_1 x + c_2 x^2 + ... + c_n x^n$

We construct the Vandermonde matrix $\mathbf{V}$:

$\mathbf{V} = \begin{bmatrix} 1 & x_1 & x_1^2 & \cdots & x_1^n \\ 1 & x_2 & x_2^2 & \cdots & x_2^n \\ \vdots & & & & \vdots \\ 1 & x_m & x_m^2 & \cdots & x_m^n \end{bmatrix}$ 

The least squares solution minimises $\|\mathbf{Vc} - \mathbf{y}\|^2$ solved via the normal equations:

$\mathbf{V}^T \mathbf{V} \, \mathbf{c} = \mathbf{V}^T \mathbf{y}$
$\mathbf{c} = (\mathbf{V}^T \mathbf{V})^{-1} \mathbf{V}^T \mathbf{y}$

#### Question 2
Maximum Likelihood (MLE) approach to curve fitting -> determine the optimal weight vector

MLE requires you to assume a data-generating process. The standard assumption is that each observation has Gaussian noise:

$y_i = c_0 + c_1 x_i + \cdots + c_4 x_i^4 + \varepsilon_i$, $\qquad \varepsilon_i \sim \mathcal{N}(0, \sigma^2)$ 

So each $y_i$ is normally distributed:

$y_i \sim \mathcal{N}(\mathbf{v}_i^T \mathbf{c},\ \sigma^2)$

The likelihood is the probability of observing your data given parameters $(\mathbf{c}, \sigma^2)$. Since points are independent:

$\mathcal{L}(\mathbf{c}, \sigma^2) = \prod_{i=1}^{n} \frac{1}{\sqrt{2\pi\sigma^2}} \exp\!\left(-\frac{(y_i - \mathbf{v}_i^T\mathbf{c})^2}{2\sigma^2}\right)$

Taking the log-likelihood (products become sums, easier to maximise):

$\ell(\mathbf{c}, \sigma^2) = -\frac{n}{2}\ln(2\pi\sigma^2) - \frac{1}{2\sigma^2}\sum_{i=1}^{n}(y_i - \mathbf{v}_i^T\mathbf{c})^2$

The Key Insight:

The $\sigma^2$ term is a constant with respect to $c$. So maximising $\ell$ over $c$ reduces to:

$\arg\max_{\mathbf{c}}\ \ell = \arg\min_{\mathbf{c}}\ \sum_{i=1}^{n}(y_i - \mathbf{v}_i^T\mathbf{c})^2$

Under Gaussian noise, MLE is identical to least squares. The normal equations give you the exact MLE solution. MLE becomes a different method when you change the noise model. Maximising log-likelihood over $c$ is identical to minimising squared residuals. The extra thing MLE gives you is sigma2_mle: once you have $c$, you substitute back into the log-likelihood and optimise over $\sigma^2$, which gives $\hat{\sigma}^2 = \frac{1}{n}\sum r_i^2$​. MLE divides by $n$, which systematically underestimates $\sigma^2$ because the fitted polynomial has already absorbed some of the noise. Dividing by $n - p$ (where $p=5$ parameters) corrects for this — this is the same correction used in standard linear regression.

Why MLE $\sigma^2$ is Biased?

What "Biased" Means:

An estimator is unbiased if its expected value equals the true parameter across repeated experiments:
$E[\hat{\sigma}^2] = \sigma^2$
The MLE estimator divides by $n$, and it can be proven mathematically that:

$E\left[\frac{1}{n}\sum r_i^2\right] = \frac{n-p}{n}\sigma^2$

So it systematically underestimates $\sigma^2$. Dividing by $(n−p)$ corrects this:

$E\left[\frac{1}{n-p}\sum r_i^2\right] = \sigma^2$


```python
# ── Approach 2: Analytic MLE (identical to normal equations) ─────────────────
# Maximising log-likelihood over c gives the same normal equations as least squares
def question2(df, order):
    x_plot = np.linspace(df.index.min(), df.index.max(), 10)
    V, w, sigma2_mle = mle_analytic(x_plot, df["y"], order)
    print("Weights = " + str(w))
    print("Sigma^2 = " + str(sigma2_mle))

    plt.figure(figsize=(10, 6))
    plt.scatter(df.index, df["y"], s=20, alpha=0.5, color='gray', label='Noisy data')
    plt.plot(x_plot, V @ w,  'r-',  lw=2,   label='MLE Approach')
    # plt.plot(x_plot, V_plot @ coeffs_numerical, 'b--', lw=2,   label='Numerical MLE (Gaussian)')
    # plt.plot(x_plot, V_plot @ coeffs_laplace,   'g:',  lw=2.5, label='Numerical MLE (Laplace / L1)')
    plt.legend()
    plt.title(f"Polynomial Fitting {order}th Order: MLE")
    plt.xlabel("x"); plt.ylabel("y")
    plt.tight_layout()
    plt.show()

if __name__ == "__main__":
    order = 4
    question2(df, order)
    order = 9
    question2(df, order)
    
```

#### Question 3
Bayesian approach to curve fitting -> Instead of asking "what are the best coefficients?", Bayesian inference asks "given the data, what should I believe about the coefficients?"

The Math:

Bayes' theorem applied to curve fitting:

${p(\mathbf{c}, \sigma \mid \mathbf{y})}_{\text{posterior}} = \frac{\underbrace{p(\mathbf{y} \mid \mathbf{c}, \sigma)}_{\text{likelihood}} \cdot \underbrace{p(\mathbf{c}, \sigma)}_{\text{prior}}}{\underbrace{p(\mathbf{y})}_{\text{evidence}}}$

Likelihood — same Gaussian model as MLE: $p(\mathbf{y} \mid \mathbf{c}, \sigma) = \mathcal{N}(\mathbf{Vc}, \sigma^2 \mathbf{I})$

- Prior — your belief about coefficients before seeing data

- Posterior — updated belief after seeing data

- Evidence — a normalising constant, usually ignored

Analytic (Conjugate Gaussian Prior):

If you assume $p(\mathbf{c}) = \mathcal{N}(\mathbf{0}, \alpha^2 \mathbf{I})$ and fix $\sigma$, the posterior has a closed form:
$p(\mathbf{c} \mid \mathbf{y}) = \mathcal{N}(\mathbf{m}, \mathbf{S})$

$\mathbf{S} = \left(\frac{1}{\sigma^2}\mathbf{V}^T\mathbf{V} + \frac{1}{\alpha^2}\mathbf{I}\right)^{-1}, \qquad \mathbf{m} = \frac{1}{\sigma^2}\mathbf{S}\mathbf{V}^T\mathbf{y}$ 

$m$ is the posterior mean (your best single estimate), and $S$ is the posterior covariance (your uncertainty).

What the Prior Does:

The prior on coefficients acts as regularisation. A narrow prior (small $\alpha$) pulls coefficients toward zero — identical in effect to Ridge regression:

$\underbrace{\text{Bayesian MAP}}_{\text{posterior mode with Gaussian prior}} \equiv \underbrace{\min_\mathbf{c} \|\mathbf{Vc} - \mathbf{y}\|^2 + \frac{\sigma^2}{\alpha^2}\|\mathbf{c}\|^2}_{\text{Ridge regression}}$ 

Zero-Mean Gaussian Prior:

When you write:
$p(\mathbf{c}) = \mathcal{N}(\mathbf{0}, \alpha^2 \mathbf{I})$

You are saying: before seeing any data, I believe each coefficient is most likely zero, and values far from zero are increasingly unlikely.

What It Looks Like:

For a single coefficient $c_i$​:

$p(c_i) = \frac{1}{\sqrt{2\pi\alpha^2}} \exp\left(-\frac{c_i^2}{2\alpha^2}\right)$

What It Means Physically:

For your degree-4 polynomial $y = c_0 + c_1 x + c_2 x^2 + c_3 x^3 + c_4 x^4$, the prior is saying:

- Before seeing data, your best guess is that the polynomial is flat — $y \approx 0$ everywhere

- You have no reason to prefer positive over negative coefficients — the prior is symmetric around zero
You believe the polynomial is probably simple — large coefficients that create wild swings are unlikely. 

Connection to Regularisation:

The zero-mean Gaussian prior is not just a philosophical choice — it has a direct mathematical consequence. Taking the log:

$\log p(\mathbf{c}) = -\frac{1}{2\alpha^2}\|\mathbf{c}\|^2 + \text{const}$

When you compute the MAP estimate, you maximise log-likelihood + log-prior:

$\mathbf{c}^{\text{MAP}} = \arg\max_\mathbf{c} \underbrace{\left[-\frac{1}{2\sigma^2}\|\mathbf{y} - \mathbf{Vc}\|^2\right]}_{\text{log-likelihood}} + \underbrace{\left[-\frac{1}{2\alpha^2}\|\mathbf{c}\|^2\right]}_{\text{log-prior}}$ 

Which is equivalent to minimising:

$\|\mathbf{y} - \mathbf{Vc}\|^2 + \underbrace{\frac{\sigma^2}{\alpha^2}}_{\lambda}\|\mathbf{c}\|^2$ 

This is exactly **Ridge regression** with regularisation strength $\lambda = \sigma^2/\alpha^2$. The zero-mean Gaussian prior *is* Ridge regularisation, just viewed through a probabilistic lens. The prior pulls coefficients toward zero — which is exactly what the $\lambda \|\mathbf{c}\|^2$ penalty term does.

```python
# ── Approach 3: Bayesian Approach ─────────────────
def question3(df, order):
    alpha = 0.005      # prior std  — how wide our prior belief is
    sigma = 1/11.1      # noise std  — assumed known (= TRUE_SIGMA here)
    m, S, V_plot, c_map = bayesian_approach(df.index, df["y"], order, alpha, sigma)

    post_var = np.diag(S)          # marginal variance per coefficient
    # post_std = np.sqrt(post_var)   # marginal std per coefficient
    n_samples = len(df.index)
    x_plot = np.linspace(df.index.min(), df.index.max(), n_samples)

    coeff_samples = np.random.multivariate_normal(m, S, size=n_samples)
    curve_samples = V_plot @ coeff_samples.T

    mean_curve = V_plot @ m
    param_var = np.array([v @ S @ v for v in V_plot])   # parameter uncertainty
    pred_var = sigma**2 + param_var                     # total predictive variance
    pred_std = np.sqrt(pred_var)
    c_mle = np.linalg.solve(V_plot.T @ V_plot, V_plot.T @ df["y"])
    mle_curve = V_plot @ c_mle
    
    # --- Panel A: Fit + credible interval ---
    plt.figure(figsize=(10, 6))
    for i in range(0, n_samples, 6):
        plt.plot(x_plot, curve_samples[:, i], color='steelblue', alpha=0.04, lw=0.8)
    plt.fill_between(x_plot,
                        mean_curve - 1.96*pred_std,
                        mean_curve + 1.96*pred_std,
                        alpha=0.18, color='steelblue', label='95% predictive CI')
    plt.fill_between(x_plot,
                        mean_curve - 1.96*np.sqrt(param_var),
                        mean_curve + 1.96*np.sqrt(param_var),
                        alpha=0.28, color='royalblue', label='95% parameter CI')
    # ax_fit.plot(x_plot, V_plot @ np.array(TRUE_COEFFS), 'k--', lw=1.5, label='true polynomial')
    plt.plot(x_plot, mle_curve,   color='tomato',    lw=1.5, ls=':', label='MLE (no prior)')
    plt.plot(x_plot, mean_curve,  color='royalblue', lw=2.0, label='posterior mean = MAP')
    plt.scatter(x_plot, df["y"], s=18, color='gray', alpha=0.6, zorder=5, label='noisy data')
    plt.legend()
    plt.title(f"Polynomial Fitting {order}th Order: Bayesian Approach (zero-mean Gaussian prior)")
    plt.xlabel('x')
    plt.ylabel('y')
    plt.show()

if __name__ == "__main__":
    order = 4 
    question3(df, order)
    order = 9
    question3(df, order)
```

#### Question 6
Bayesian Model Selection -> determining most probable model order. Plotting marginal likelihood vs model order for the dataset. 

The core idea — marginal likelihood:

The key quantity is $p(\mathbf{y} \mid M_k)$ — the probability of observing your data under model order $k$, with the coefficients integrated out:

$p(\mathbf{y} \mid M_k) = \int p(\mathbf{y} \mid \mathbf{c}, M_k)\; p(\mathbf{c} \mid M_k)\; d\mathbf{c}$

Because both terms are Gaussian, this integral has a closed form. The marginal covariance is $\mathbf{C} = \sigma^2\mathbf{I} + \alpha^2 \mathbf{V}\mathbf{V}^T$, but computing its determinant at size $(60×60)$ is expensive. The code instead uses the equivalent posterior-based form:

$p(\mathbf{y} \mid M_k) = \underbrace{-\frac{N}{2}\log(2\pi\sigma^2)}_{\text{constant}} \underbrace{-\frac{p}{2}\log\alpha^2 + \frac{1}{2}\log|\mathbf{S}|}_{\text{prior term}} \underbrace{-\frac{\|\mathbf{y}\|^2}{2\sigma^2}}_{\text{data fit}} \underbrace{+\frac{\mathbf{m}^T \mathbf{S}^{-1} \mathbf{m}}{2\sigma^2}}_{\text{posterior gain}}$

The math:

Instead of working with the posterior quantities $S$ and $m$, go back to the original marginal distribution directly:

$p(\mathbf{y} \mid M) = \mathcal{N}(\mathbf{y}; \mathbf{0}, \mathbf{C})$ where the marginal covariance is:

$\mathbf{C} = \sigma^2 \mathbf{I}_N + \alpha^2 \mathbf{V}\mathbf{V}^T \qquad \text{shape: } (N \times N)$

The log marginal likelihood is then:

$\log p(\mathbf{y} \mid M) = -\frac{N}{2}\log(2\pi) - \frac{1}{2}\log|\mathbf{C}| - \frac{1}{2}\mathbf{y}^T \mathbf{C}^{-1} \mathbf{y}$

Just a direct matrix inverse and determinant on the full $N×N$ system.

#### Question 7
Performing Bayesian Model averaging -> plotting the predictive mixture distribution mean and standard deviation

The BMA predictive distribution mixes all models weighted by their posterior probabilities:

$p(y^* \mid x^*, \mathbf{y}) = \sum_k p(M_k \mid \mathbf{y}) \cdot \mathcal{N}(y^* ; \mu_k, \sigma_k^2)$

This is a Gaussian mixture — not itself Gaussian. The mean and variance follow from the law of total expectation and total variance.

```python
def question7(df, N, alpha, sigma):
    orders = list(range(0, 10))   # test orders 0 through 9
    log_mls = []
    for o in orders:
        log_ml, m, S, V = log_marginal_likelihood(N, df.index, df["y"], o, alpha, sigma)
        log_mls.append(log_ml)

    # Softmax for numerical stability
    log_mls_shifted = log_mls - np.max(log_mls)
    unnorm = np.exp(log_mls_shifted)
    model_weights = unnorm/unnorm.sum()

    n_samples = len(df.index)
    x_plot = np.linspace(df.index.min(), df.index.max(), n_samples)
    
    per_model = {}
    for o in orders:
        m_k, S_k, _ = compute_posterior(df.index, df["y"], o, alpha, sigma)
        V_plot = build_vandermonde(x_plot, o)

        mu_k = V_plot @ m_k
        var_k = sigma**2 + np.array([v @ S_k @ v for v in V_plot])
        per_model[o] = {'mu': mu_k, 'var': var_k, 'weight': model_weights[o]}

    bma_mean = sum(per_model[o]['weight'] * per_model[o]['mu']
               for o in orders)                                 # (300,)
 
    # within-model variance (aleatoric + parameter uncertainty per model)
    within_var = sum(per_model[o]['weight'] * per_model[o]['var']
                    for o in orders)                               # (300,)
    
    # between-model variance (model disagreement — epistemic)
    between_var = sum(per_model[o]['weight'] * (per_model[o]['mu'] - bma_mean)**2
                    for o in orders)                              # (300,)
    
    bma_var = within_var + between_var                             # total variance
    bma_std = np.sqrt(bma_var)                                     # total std
    
    within_std  = np.sqrt(within_var)
    between_std = np.sqrt(between_var)

    # ── Panel B: individual model predictive means (weighted by opacity)
    plt.figure(figsize=(10, 6))
    cmap = plt.cm.viridis(np.linspace(0.1, 0.9, len(orders)))
    for o in orders:
        w   = per_model[o]['weight']
        mu  = per_model[o]['mu']
        std = np.sqrt(per_model[o]['var'])
        if w > 1e-4:
            plt.plot(x_plot, mu, color=cmap[o], lw=1.5, alpha=min(1.0, w*4 + 0.1),
                        label=f'order {o}  (w={w:.3f})')
            plt.fill_between(x_plot, mu - std, mu + std,
                                color=cmap[o], alpha=w * 0.8)
    plt.scatter(df.index, df["y"], s=15, color='gray', alpha=0.4, zorder=5)
    plt.title('Per-model predictive distributions\n(opacity ∝ model weight)')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.legend()
    plt.show()

    plt.figure(figsize=(10, 6))
    plt.fill_between(x_plot,
                        bma_mean - 2*bma_std,
                        bma_mean + 2*bma_std,
                        alpha=0.15, color='steelblue', label='BMA ±2 std (total)')
    plt.fill_between(x_plot,
                        bma_mean - bma_std,
                        bma_mean + bma_std,
                        alpha=0.30, color='steelblue', label='BMA ±1 std (total)')
    plt.plot(x_plot, bma_mean,  color='steelblue', lw=2.5, label='BMA mean')
    plt.scatter(df.index, df["y"], s=18, color='gray', alpha=0.5, zorder=5, label='data')

    # overlay best single model for comparison
    best_order = orders[np.argmax(model_weights)]
    plt.plot(x_plot, per_model[best_order]['mu'],
                color='tomato', lw=1.5, ls='--',
                label=f'best single model (order {best_order})')
    plt.title('Bayesian model averaging — predictive mixture distribution\n'
                    'E[y*] = Σ_k w_k μ_k(x*)     Var[y*] = E[Var] + Var[E]')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.legend()
    plt.show()

if __name__ == "__main__":
    question7(df, N, alpha, sigma)
 
```

#### Question 3
Feature types -> investigating the Hog Feature Extraction Algorithm:

HOG (Histogram of Oriented Gradients) captures the shape and edge structure of an image by measuring which direction edges point across small regions. Instead of raw pixel values, HOG asks: "in this small patch of the image, which directions do the edges run?" A '7' has strong horizontal edges at the top and diagonal edges going down-right. A '1' has mostly vertical edges. These edge direction patterns are distinctive and stable across different handwriting styles.

Other types of Feature Extraction include Loops, auch as Flood Fill, Extracting Endpoints, and extracting the Center of Mass.

```python
def plot_hog_gradients(image, cell_size=7, n_bins=9):
    image = image.reshape(28, 28)

    # Compute gradients
    gx = np.zeros_like(image)
    gy = np.zeros_like(image)
    gx[:, 1:-1] = image[:, 2:] - image[:, :-2]
    gy[1:-1, :] = image[2:, :] - image[:-2, :]
    magnitude = np.sqrt(gx**2 + gy**2)
    angle     = np.arctan2(gy, gx) * 180 / np.pi % 180

    fig, axes = plt.subplots(1, 4, figsize=(16, 4))

    axes[0].imshow(image,     cmap='gray')
    axes[0].set_title('Original')

    axes[1].imshow(gx,        cmap='RdBu')
    axes[1].set_title('Gradient X')

    axes[2].imshow(gy,        cmap='RdBu')
    axes[2].set_title('Gradient Y')

    axes[3].imshow(magnitude, cmap='hot')
    axes[3].set_title('Gradient Magnitude')

    for ax in axes:
        ax.axis('off')

    plt.suptitle('HOG Gradient Decomposition')
    plt.tight_layout()
    plt.show()

if __name__ == "__main__":
    x_train_raw, X_train, Y_train, X_test, Y_test = load_data()
    plot_hog_gradients(X_test[0])
```

#### Question 4
The Naive Bayesian Classifier

A Bayesian classifier is a probabilistic model that uses Bayes' theorem to predict the class of a data point. It calculates the probability that a data point belongs to each class, then assigns it to the most likely one.
Bayes' theorem states:

P(Class | Features) = P(Features | Class) × P(Class) / P(Features)

Naive Bayes says that each data point is independent from every other data point 

Where:

- P(Class | Features) — posterior: probability of the class given the observed features
- P(Features | Class) — likelihood: how probable the features are under that class
- P(Class) — prior: baseline frequency of the class in training data
- P(Features) — evidence: a normalizing constant (same for all classes, so often ignored)

1. Estimate priors P(Class):

    Count the frequency of each class in the training set:

    P(Class = spam) = num_spam_emails / total_emails

2. Estimate likelihoods P(fᵢ | Class):

    For each feature, estimate its probability within each class. The method depends on the feature type:

    Categorical features — use frequency counts (e.g. word present/absent)

    Continuous features — fit a Gaussian (mean + variance) per class

    P("free" | spam)     = count("free" in spam) / count(spam)

    P("free" | not_spam) = count("free" in not_spam) / count(not_spam)

    Important: Apply Laplace (add-1) smoothing to avoid zero probabilities for unseen feature values:
    
    P(word | class) = (count + 1) / (total_words_in_class + vocabulary_size)

3. Classify new data points

    For a new sample with features [f₁, f₂, ..., fₙ], compute the score for each class and pick the highest:

    score(Class) = P(Class) × P(f₁|Class) × P(f₂|Class) × ... × P(fₙ|Class)
    
    In practice, use log probabilities to avoid floating-point underflow:
    
    log_score(Class) = log P(Class) + Σ log P(fᵢ | Class)

The Core Limitation
The independence assumption breaks down hardest with images because a pixel's value is strongly predicted by its neighbors. This is why:

Naive Bayes on raw pixels works for simple, aligned datasets (MNIST, fixed-pose faces)
It fails for complex scenes, rotation, scale changes, or cluttered backgrounds
CNNs (Convolutional Neural Networks) were specifically designed to exploit spatial correlation — they are the natural successor for image tasks

#### Question 5
Implementation of the Naive Bayesian Classifier

Two types of Naive Bayesian classifiers are explored here: Gaussian Naive Bayes, and Bernoulli Naive Bayes. It is evident that the Bernoulli implementation is more accurate than the Gaussian Method, as Bernoulli applies Binarization by creating a threshold for each pixel. This works well for high-contrast images such as handwritten text.  

```python
# ── Gaussian ──────────────────────────────────────────────────────────────────────
def gaussian_naive_bayes(X_train, Y_train):
    classes = np.unique(Y_train)
    priors  = np.zeros(10)
    means   = np.zeros((10, 784))
    variances = np.zeros((10, 784))

    for i, c in enumerate(classes):
        X_c = X_train[Y_train == c]
        priors[i] = X_c.shape[0]/len(X_train)
        means[i] = X_c.mean(axis=0)
        variances[i] = X_c.var(axis=0) + 1e-9

    return priors, means, variances

def predict_gaussian(X, priors, means, variances):
    scores = np.zeros((len(X), 10))
    for i in range(10):
        scores[:, i] = (
            np.log(priors[i])
            - 0.5 * np.sum(np.log(2 * np.pi * variances[i]))
            - 0.5 * np.sum(((X - means[i]) ** 2) / variances[i], axis=1)
        )
    return np.argmax(scores, axis=1)

# ── Bernoulli ────────────────────────────────────────────────────────────────────
def bernoulli(X_train, Y_train):
    probs = np.zeros((10, 784))
    priors = np.zeros(10)

    for i, c in enumerate(range(10)):
        X_c = X_train[Y_train == c]
        priors[i] = X_c.shape[0] / len(X_train)
        probs[i]  = (X_c.sum(axis=0) + 1) / (X_c.shape[0] + 2)   # Laplace smoothing
    return probs, priors

def predict_bernoulli(X, probs, priors):
    scores = np.zeros((len(X), 10))
    for i in range(10):
        scores[:, i] = (
            np.log(priors[i])
            + X @ np.log(probs[i])              # pixel ON contribution
            + (1 - X) @ np.log(1 - probs[i])   # pixel OFF contribution
        )
    return np.argmax(scores, axis=1)


if __name__ == "__main__":
    x_train_raw, X_train, Y_train, X_test, Y_test = load_data()
    
    # ── Gaussian ───────────────────────────────────────────────────────────────────
    # Train data
    priors, means, variances = gaussian_naive_bayes(X_train, Y_train)
    # Test data
    y_pred = predict_gaussian(X_test, priors, means, variances)
    print(f"Gaussian NB Accuracy: {np.mean(y_pred == Y_test):.2%}")

    # Confusion Matrix for testing 
    cm = np.zeros((10, 10), dtype=int)

    for true, pred in zip(Y_test, y_pred):
        cm[true][pred] += 1

    plt.figure(figsize=(10, 8))
    plt.imshow(cm, cmap='Blues')
    plt.colorbar()
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.title('Confusion Matrix Gaussian')
    for i in range(10):
        for j in range(10):
            plt.text(j, i, str(cm[i][j]), ha='center', va='center',
                    color='white' if cm[i][j] > cm.max()/2 else 'black')
    plt.tight_layout()
    plt.show()

    # ── Bernoulli ───────────────────────────────────────────────────────────────────
    X_bin_train = (X_train > 0.5).astype(float)
    X_bin_test = (X_test > 0.5).astype(float)

    # Train data
    probs, priors = bernoulli(X_bin_train, Y_train)
    # Test data
    y_pred = predict_bernoulli(X_bin_test, probs, priors)
    print(f"NB Bernoulli Accuracy: {np.mean(y_pred == Y_test):.2%}")   # ~84%

    # Confusion Matrix for testing 
    cm = np.zeros((10, 10), dtype=int)

    for true, pred in zip(Y_test, y_pred):
        cm[true][pred] += 1

    plt.figure(figsize=(10, 8))
    plt.imshow(cm, cmap='Blues')
    plt.colorbar()
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.title('Confusion Matrix Bernoulli')
    for i in range(10):
        for j in range(10):
            plt.text(j, i, str(cm[i][j]), ha='center', va='center',
                    color='white' if cm[i][j] > cm.max()/2 else 'black')
    plt.tight_layout()
    plt.show()
```
