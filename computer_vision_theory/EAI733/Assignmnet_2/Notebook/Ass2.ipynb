### 1. Camera Calibration

### 1.1 Overview of camera geometry and camera parameters
#### Intrinsic vs Extrinsic Camera Parameters in Computer Vision

Extrinsic parameters of a camera depend on location and orientation within an environment, while the intrinsic parameters depend on how images are captured, such as focal length, aperture, field-of-view, and resolution. They are transformation matrices that convert points from one co-ordinate system to another: extrinsic is from world to camera, and intrinsic is from camera to pixel.

Intrinsic and extrinsic camera calibration allows for a 3D representastion of the world to be converted to a 2D projection as an image, allowing for image processing. 

### 1.2 Review of auto-calibration techniques
In photogrammetric calibration, camera calibration is performed by observing a calibration object whose geometry in 3D space is known with very good precision, however expensive calibration apparatus and an elaborate set-up is needed. 

In self-calibration, a camera is moved in a static scene, and the rigidity of the scene creates constraints on the camera's internal parameters. Thus, after taking several images with these fixed constraints, corresepondences between the images is sufficient to recover the internal and external parameters. There must be suffiecient displacement between the images (taken from different camnera angles) so that there is enough independent data to accurately solve for the camera parameters.

Techniques include tracking points over time, structure from motion or using LiDAR to minimise reprojection errors.
